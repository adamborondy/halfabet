```python
import pandas as pd
import json

df = pd.read_excel('OpenLexicon.xlsx')
```

```python
df.head()
```

```python

df = df[df['ortho'].str.match(r'^[a-zA-Z]+$')]
df['word'] = df['ortho'].str.lower()
df['freq'] = df['SUBTLEXus__FREQcount']

# Drop duplicates, sort by frequency
df = df[['word', 'freq']].drop_duplicates('word').sort_values('freq')

# Assign percentile 0–100
df['percentile'] = (df['freq'].rank(pct=True) * 100).round(1)
```

```python
df
```

```python
# Convert to dict and save
result = dict(zip(df['word'], df['percentile']))
with open('words.json', 'w') as f:
    json.dump(result, f)

print(f"Done — {len(result)} words")
```
